# Finite automaton drawn with graphviz

```python
from graphviz import Digraph

def create_automaton():
    # Define the Digraph
    f = Digraph('finite_state_machine', filename='fsm.gv')
    f.attr(rankdir='LR', size='8,5')

    # Define the states and the transitions
    f.attr('node', shape='doublecircle')
    f.node('q2')

    f.attr('node', shape='circle')
    f.node('q0')
    f.node('q1')

    f.edge('q0', 'q1', label='identifiant')
    f.edge('q1', 'q2', label='opEgal')
    f.edge('q2', 'q2', label='nombre')
    f.edge('q2', 'q2', label='identifiant')
    f.edge('q2', 'q2', label='operateur')

    return f

# Create the automaton graph
fsm = create_automaton()
# Render the graph to a file
fsm.render('fsm_output', view=True, format='pdf')

# Return the path to the rendered graph file
fsm_file_path = fsm.filepath + '.pdf'
fsm_file_path
```
