# OEIL MAGIQUE
## Rajouter une troisième dimension hue

Le paramètre **hue** peut être utilisé dans la quasi-totalité des commandes vues précédemment permet de rajouter une 3ème dimension aux variables étudiées

<u>Exemples :</u>

```python
sns.relplot(data = tips,
               x = 'total_bill',
               y = 'tip',
               hue = 'smoker',
               kind = 'scatter')
```

```python
# paramétrage des couleurs
# les couleurs peuvent être en HTML
couleurs = { 'Yes' : 'red',
               'No' : 'blue'}
sns.relplot(data = tips,
               x = 'total_bill',
               y = 'tip',
               hue = 'smoker',
               palette = couleurs,
               kind = 'scatter')
```
