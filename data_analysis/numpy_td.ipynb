Les fonctions **zeros** ou **ones** créant respectivement un tableau de

Si on veut faire subir une rotation de 180° à l'image, il suffit de faire la commande suivante de PIL :

Si on veut transformer l'image en un tableau numpy, il suffit d'utiliser la commande« magique » array de numpy :

```python
import numpy as np
image_np = np.array(image)
```

On vérifie le type et le format de cette variable :

Expliquez chaque ligne

```python
print(type(image_np))
print(image_np.dtype)
print(image_np.shape)
```

L'image corresponde donc à une matrice à 3 dimensions, les deux
premières dimensions correspondant à la hauteur et la largeur de
l'image, la 3ème dimension correspondant à 3 valeur (R,G,B) qui sont les
composantes rouge, verte et bleu.

### Exercice (manipulation de tranches de numpy) :

Transformez l'image originale, en ne conservant que la composante rouge
pour obtenir l'image suivante :

```python
image_rs = image_np.copy()
image_b = image_np.copy()
image_r = image_np.copy()
image_rs[:,:,(1,1)] = 0
image_b[:,:,(0,1)] = 0
image_r[:,:,(1,2)] = 0
test_v = Image.fromarray(image_rs)
test_v
```

Pour ce faire, vous pourrez utiliser les commandes suivantes :

* méthode copy pour copier et garder l'image originale image_np

* Passez les composantes verte et bleue à 0 pour ne garder que la
composante rouge.

* Transformez votre « array » numpy en image avec la méthode
Image.fromarray()
