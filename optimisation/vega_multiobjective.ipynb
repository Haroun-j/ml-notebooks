# Algorithme VEGA

## Objectifs :
#### 1. Maximiser la publicité totale
#### 2. Minimiser le salaire total
#### 3. Maximiser le nombre de célébrités
# ## Contrainte :
#### Poids total ≤ 1000 kg
# ## Données issues du TD :
#### 8 célébrités avec (salaire, publicité, poids)
#### Population initiale : 6 individus donnés

```python

import random

# Données
celebrites = [
    (100, 10, 50),
    (200, 20, 70),
    (50, 5, 80),
    (600, 60, 55),
    (400, 40, 67),
    (300, 30, 59),
    (500, 50, 74),
    (700, 70, 90)
]

# Capacité du sac
CAPACITE = 1000
NB_OBJETS = len(celebrites)

# Population initiale
population = [
    "10010110",  # S1
    "11000011",  # S2
    "11110000",  # S3
    "01011010",  # S4
    "10000001",  # S5
    "00001000"   # S6
]

# Fonctions d’évaluation
# Salaire total d'une solution
def calcul_salaire(sol):
    return sum(int(sol[i]) * celebrites[i][0] for i in range(NB_OBJETS))

# Publicité totale
def calcul_publicite(sol):
    return sum(int(sol[i]) * celebrites[i][1] for i in range(NB_OBJETS))

# Poids total
def calcul_poids(sol):
    return sum(int(sol[i]) * celebrites[i][2] for i in range(NB_OBJETS))

# Nbr total de célèbrité
def nb_celebrite(sol):
    return sol.count("1")

# Contrainte de poids
def est_faisable(sol):
    return calcul_poids(sol) <= CAPACITE

# Sélection VEGA
def selection_vega(pop):
    par_pub = sorted(pop, key=calcul_publicite, reverse=True)  # max pub
    par_sal = sorted(pop, key=calcul_salaire)                  # min salaire
    par_nb  = sorted(pop, key=nb_celebrite, reverse=True)      # max nb

    sel = par_pub[:2] + par_sal[:2] + par_nb[:2]
    return sel

# Croisement à un point entre deux solutions
def croisement(p1, p2):
    point = random.randint(1, NB_OBJETS - 1)
    enfant1 = p1[:point] + p2[point:]
    enfant2 = p2[:point] + p1[point:]
    return enfant1, enfant2

# Mutation : inversion d’un bit avec une faible probabilité
def mutation(sol, proba=0.05):
    s = list(sol)
    for i in range(NB_OBJETS):
        if random.random() < proba:
            s[i] = '1' if s[i] == '0' else '0'
    return "".join(s)

# Exécution de l’algorithme VEGA
NB_GENERATIONS = 10
random.seed(5)
print("=== Population initiale ===")
for s in population:
    print(s, "| Salaire:", calcul_salaire(s),
              "| Pub:", calcul_publicite(s),
              "| Nb:", nb_celebrite(s),
              "| Poids:", calcul_poids(s),
              "| Faisable:", est_faisable(s))

for gen in range(NB_GENERATIONS):
    parents = selection_vega(population)

    # Création de nouveaux individus
    nouvelle_pop = []
    random.shuffle(parents)
    for i in range(0, len(parents), 2):
        p1, p2 = parents[i], parents[(i+1) % len(parents)]
        e1, e2 = croisement(p1, p2)
        e1, e2 = mutation(e1), mutation(e2)
        nouvelle_pop += [e1, e2]

    # Mise à jour
    population = nouvelle_pop[:6]
    print(f"\n--- Génération {gen+1} ---")
    for s in population:
        print(s, "| Salaire:", calcul_salaire(s),
                  "| Pub:", calcul_publicite(s),
                  "| Nb:", nb_celebrite(s),
                  "| Poids:", calcul_poids(s))

# Extraction du front de Pareto
def domine(a, b):
    return (
        calcul_publicite(a) >= calcul_publicite(b) and
        calcul_salaire(a) <= calcul_salaire(b) and
        nb_celebrite(a) >= nb_celebrite(b) and
        (
            calcul_publicite(a) > calcul_publicite(b) or
            calcul_salaire(a) < calcul_salaire(b) or
            nb_celebrite(a) > nb_celebrite(b)
        )
    )
```
