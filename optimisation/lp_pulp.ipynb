# Linear programming: PuLP model and graphical feasible regions

## Integer program solved with PuLP

```python
import pulp

# Création du modèle
model = pulp.LpProblem("Maximiser_z", pulp.LpMaximize)

# Définition des variables (entiers non-négatifs)
x1 = pulp.LpVariable('x1', lowBound=0, cat='Integer')
x2 = pulp.LpVariable('x2', lowBound=0, cat='Integer')

# Fonction objectif
model += x1 + 2 * x2, "Z"

# Contraintes
model += -3 * x1 + 2 * x2 <= 2
model += -x1 + 2 * x2 <= 4
model += 3 * x1 + 2 * x2 <= 15

# Résolution du modèle
model.solve()

# Affichage de la solution
if pulp.LpStatus[model.status] == 'Optimal':
    print("Solution optimale:")
    print(f"x1 = {x1.value()}")
    print(f"x2 = {x2.value()}")
    print(f"Valeur maximale de z = {pulp.value(model.objective)}")
else:
    print("Aucune solution optimale trouvée.")
```

## Feasible regions of small linear programs

```python
import numpy as np
import matplotlib.pyplot as plt

# Définir les limites du graphique
x1 = np.linspace(0, 5, 100)
x2 = np.linspace(0, 5, 100)
x1, x2 = np.meshgrid(x1, x2)

# Définir les inégalités
cond1 = 2*x1 + 4*x2 <= 10
cond2 = -3*x1 + 4*x2 <= -2
cond3 = -x1 - x2 <= 0

# Plot
plt.figure()
plt.title('A1')
plt.xlabel('x1')
plt.ylabel('x2')
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.contourf(x1, x2, (cond1 & cond2 & cond3), levels=[0, 1], colors=['blue'], alpha=0.5)
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Définir les limites pour x1 et x2
x1_values = np.linspace(-10, 10, 400)
x2_values = np.linspace(-10, 10, 400)

# Créer une grille de valeurs pour x1 et x2
x1_grid, x2_grid = np.meshgrid(x1_values, x2_values)

# Définir l'inégalité
inequality = 2*x1_grid + 4*x2_grid <= 10

# Afficher la région où l'inégalité est satisfaite
plt.imshow(inequality, extent=(-10, 10, -10, 10), origin='lower', cmap='Greys', alpha=0.3)

# Afficher l'inégalité comme une ligne
plt.plot(x1_values, (10 - 2*x1_values) / 4, label='2*x1 + 4*x2 = 10', color='red')

# Ajouter des étiquettes et une légende
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Inégalité 2*x1 + 4*x2 <= 10')
plt.legend()

# Afficher le graphique
plt.grid(True)
plt.show()
```
